```python
import numpy as np
from matplotlib import pyplot as plt

from common.rewards import MazeRLReward
```

```python
rho = np.zeros(shape=(10, ), dtype=float) + 0.5

for num_disconnected in range(0, 3):
    rho_maxs = np.linspace(0, 1.5, 50)
    rewards = []
    for rho_max in rho_maxs:
        rho[0] = rho_max
        rew = MazeRLReward.rew(num_disconnected, rho)
        rewards.append(rew)

    plt.plot(rho_maxs, rewards, label=f"num_disconnected={num_disconnected}")

plt.legend(loc="best")
plt.xlabel("rho_max")
plt.ylabel("reward")
plt.title("Reward over rho_max")
plt.show()

```

```python
rho = np.zeros(shape=(10, ), dtype=float)

for num_disconnected in range(0, 3):
    rho_maxs = np.linspace(0, 1.5, 50)
    rewards = []
    for rho_max in rho_maxs:
        rho[[0,1,2]] = rho_max
        rew = MazeRLReward.rew(num_disconnected, rho)
        rewards.append(rew)

    plt.plot(rho_maxs, rewards, label=f"num_disconnected={num_disconnected}")

plt.legend(loc="best")
plt.xlabel("rho_max")
plt.ylabel("reward")
plt.title("Reward over rho_max when multiple rhos grow")
plt.show()

```
