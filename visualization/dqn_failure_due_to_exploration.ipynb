```python
import numpy as np
from matplotlib import pyplot as plt

failure_on_random_action = 0.5
random_action_prob_per_step = 0.07
failure_chance_per_step = failure_on_random_action * random_action_prob_per_step
n = np.arange(30)
y = 1 - np.pow(1 - failure_chance_per_step, n)

plt.plot(n, y)
plt.xlabel("Number of steps")
plt.ylabel("Failure probability due to exploration")
```
