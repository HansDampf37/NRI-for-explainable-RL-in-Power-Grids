```python
import numpy as np
import matplotlib.pyplot as plt

# α from 0 to 1
alphas = np.linspace(0, 1, 200)
# numerical epsilons used for clipping
epsilons = np.logspace(-50, -2, 100)

A, E = np.meshgrid(alphas, epsilons)

# p = [α, 1−α], q = [0, 1]
p1 = A
p2 = 1 - A
q1 = 0
q2 = 1

# compute KL with added epsilon for numerical stability
KL = p1 * (np.log(p1 + E) - np.log(q1 + E)) + p2 * (np.log(p2 + E) - np.log(q2 + E))

# plot surface
fig = plt.figure(figsize=(6, 5))
ax = fig.add_subplot(111, projection="3d")
ax.plot_surface(A, np.log10(E), KL, cmap="viridis")

ax.set_xlabel(r'$\alpha$')
ax.set_ylabel(r'$\log_{10}(\epsilon)$ (numerical)')
ax.set_zlabel(r'$D_{KL}([α,1-α] \parallel [0,1])$')
ax.set_title('KL Divergence vs α and numerical ε')
ax.invert_yaxis()
# plt.tight_layout()
plt.show()
```
