```python
from tempfile import TemporaryDirectory

import numpy as np
from tqdm import tqdm

from src.common.env import G2OpGymEnv

env = G2OpGymEnv()

episode_lengths = []
interactions = []

for i in tqdm(range(100)):
    env.reset()
    done = False
    length = 0
    while not done:
        obs, rew, done, trunc, info = env.step(0) # do nothing
        length += 1

    episode_lengths.append(env.episode_lengths[-1])
    interactions.append(length)

print("Episode_length: ", episode_lengths)
print("Interactions: ", interactions)

fractions = np.array(episode_lengths) / np.array(interactions)
print("Mean fraction: ", np.mean(fractions))
```
